### **Built-in structures **

Lists - are the structures that store data in order. Lists have built-in methods (see below) for adding more data, removing data, modifying data, and accessing data in the list.

Tuples - tuples are structures that store data in order but are immutable. That is tuples are like lists, except that you can't add or remove from them without creating a whole new object in memory.

Sets - sets are like lists but they are unordered and no duplicates are allowed.

Dictionary - dictionaries are structures that store objects in pairs. Each pair consists of a key and a value. In analogy, a phone book is a dictionary, where the key is a person's name, and the value their phone number.

Let's start with lists..



```python
# example 1 - creating a list of ints
a = [1,2,3,4]
print("a:"+str(a))

# example 2 - adding an element and creating a new list
b = a + [100]
print(type(b))
print("b:"+str(b))

# example 3 - removing an element from the list and storing it in a new variable
c = b.pop(0)
print("c:"+str(c)+", b:"+str(b))

# example 4 - returning the minimum value
d = min(a)
print("d:"+str(d))

# example 5 - grabbing the first item in a list (without removal)
e = a[0]
print("e:"+str(e))
```

```python
# example 6 - grabbing the last object in a list (without removal)
f = a[-1]
print("f:"+str(f))

# example 7 - grabbing the first three objects in a list (without removal)
g = a[:3]
print("g:"+str(g))

# example 8 - grabbing the last three objects in a list (without removal)
h = a[-3:]
print("h:"+str(h))

# example 9 - return all objects from the list (without removal)
i = a[:]
print("i:"+str(i))

# example 10 - grabbing every other value from a list (without removal)
j = a[::2]
print("j:"+str(j))

# example 11 - reassign the first element to 100000.0
a[0] = 100000.0
print("a:"+str(a))
```

Since it is so important, here's a quick summary of slice notation. list[start:stop:step] returns objects between the indices start through stop (excluding stop). list[start:] returns objects start through the end of the list, list[:stop] returns objects from the beginning of the list through the object at index stop (excluding the object at stop), list[:] returns all objects, and list[::step] returns objects every step apart. You can use slice notation for accessing objects from any ordered data structure (e.g., tuple).

A Better Way to Read Large Files:File objects also have defined iteration rules. This allows you to loop over them line by line without reading the whole thing into memory. This is almost always preferred:

```python
# Example 1: Create a class with only attributes (a,b,c) no methods.
class test:
  a=[1,2,3]
  b="string"
  c=1

# Create an instance of our class and print its attributes
instance = test()
print(instance)
print(instance.a)
print(instance.b)
print(instance.c)
```

Here, data is included in the class itself. How can we create a class where we can get outside data into our class instance?

```python
# Example 2: Create a class with attributes assigned upon initialization.
class test:
  def __init__(self,a,b,c):
    self.a=a
    self.b=b
    self.c=c

# Create an instance of our class and print its attributes
inst = test([1,2,3],"string",1)
print(inst.a)
print(inst.b)
print(inst.c)

inst2 = test([3,9], 4,2)
print(inst2.a)
```

The easiest way to think about classes in Python is to imagine them as blueprints for creating objects.

`__init__` : method - `__init__` is a special function that Python automatically runs whenever you create a new object.

my_car = Car("Red", "Toyota")

When Python sees this line, it automatically does:

Car.`__init__`(my_car, "Red", "Toyota")

You don't explicitly call `__init__`. Python does it for you.

First, it gives the programmer the ability to pass inputs when an instance of the class is created. Second, it gives defines a pointer to the class instance (`self` by convention) that needs to be used when creating instance specific attributes and subsequent methods.

```python
# Example 3: Create a class with a custom equals comparison
class test:
  def __init__(self,a,b,c):
    self.a=a
    self.b=b
    self.c=c

  def add(self):
    return self.a + self.b + self.c

  def __eq__(self,other):
    return self.c == other.c

inst1 = test(1,1,1)
inst2 = test(12,23,3)
inst3 = test(4,4,1)

# since 1 is the same for inst1 and inst3, it shows True.
print(inst1.add())
print(inst2.__eq__(inst3))
print(inst1 == inst3)
print(inst3.add())

print("inst1 == inst2 : {}".format(inst1 == inst2))
print("inst1 == inst3: {}".format(inst1 == inst3))
```
