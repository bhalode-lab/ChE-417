## Part 1: Series

A **Series** is a one-dimensional labeled array capable of holding any data type.

### 1.1 Creating a Series

```python
# From a list
s1 = pd.Series([10, 20, 30, 40, 50])
print("Series from list:")
print(s1)
print()

# From a list with custom index
s2 = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])
print("Series with custom index:")
print(s2)
print()

# From a dictionary
s3 = pd.Series({'NY': 8.4, 'LA': 3.9, 'Chicago': 2.7, 'Houston': 2.3})
print("Series from dictionary (city populations in millions):")
print(s3)
```

### 1.2 Index Operations

```python
# Accessing the index
print("Index of s2:", s2.index)
print("Values of s2:", s2.values)
print()

# Reindexing
s_reindexed = s2.reindex(['a', 'b', 'c', 'f', 'g'])  # 'f' and 'g' will be NaN
print("Reindexed series (note NaN for new indices):")
print(s_reindexed)
print()

# Reset index to default integers
s_reset = s2.reset_index(drop=True)
print("Reset index:")
print(s_reset)
```

### 1.3 Slicing and Selection

```python
# Label-based indexing (.loc)
print("s2.loc['c']:", s2.loc['c'])
print("s2.loc['b':'d']:\n", s2.loc['b':'d'])  # inclusive
print()

# Integer-based indexing (.iloc)
print("s2.iloc[2]:", s2.iloc[2])
print("s2.iloc[1:4]:\n", s2.iloc[1:4])  # exclusive of end
print()

# Boolean indexing (filtering)
print("Values > 25:\n", s2[s2 > 25])
print()

# Multiple conditions
print("Values between 15 and 35:\n", s2[(s2 > 15) & (s2 < 35)])
```

### 1.4 Adding and Dropping Elements

```python
# Adding a single element
s2['f'] = 60
print("After adding 'f':\n", s2)
print()

# Adding via append (creates new Series)
s_new = pd.Series({'g': 70, 'h': 80})
s_combined = pd.concat([s2, s_new])
print("Combined series:\n", s_combined)
print()

# Dropping elements
s_dropped = s_combined.drop(['a', 'h'])
print("After dropping 'a' and 'h':\n", s_dropped)
print()

# Dropping in-place
s_temp = s2.copy()
s_temp.drop('a', inplace=True)
print("After in-place drop of 'a':\n", s_temp)
```

## Part 2: DataFrames

A **DataFrame** is a two-dimensional labeled data structure with columns of potentially different types.

### 2.1 Creating a DataFrame

```python
# From a dictionary of lists
df1 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [25, 30, 35, 28, 32],
    'Department': ['CS', 'Math', 'Physics', 'CS', 'Math'],
    'GPA': [3.8, 3.5, 3.9, 3.7, 3.6]
})
print("DataFrame from dictionary:")
print(df1)
print()

# From a list of dictionaries
df2 = pd.DataFrame([
    {'A': 1, 'B': 2, 'C': 3},
    {'A': 4, 'B': 5, 'C': 6},
    {'A': 7, 'B': 8, 'C': 9}
])
print("DataFrame from list of dicts:")
print(df2)
print()

# From a 2D NumPy array with custom index and columns
df3 = pd.DataFrame(
    np.random.randn(4, 3),
    index=['row1', 'row2', 'row3', 'row4'],
    columns=['colA', 'colB', 'colC']
)
print("DataFrame from NumPy array:")
print(df3)
```

### 2.2 Index Operations

```python
# Setting a column as index
df1 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [25, 30, 35, 28, 32],
    'Department': ['CS', 'Math', 'Physics', 'CS', 'Math'],
    'GPA': [3.8, 3.5, 3.9, 3.7, 3.6]
})

df_indexed = df1.set_index('Name')
print("DataFrame with 'Name' as index:")
print(df_indexed)
print()

# Resetting index
df_reset = df_indexed.reset_index()
print("After reset_index():")
print(df_reset)
print()

# Setting a multi-level index
df_multi = df1.set_index(['Department', 'Name'])
print("Multi-level index (Department, Name):")
print()

# Accessing index and columns
print("Index:", df1.index)
print(df_multi.index)
print("Columns:", df1.columns)
print("Values shape:", df1.values.shape)
```

### 2.3 Slicing and Selection

```python
# Selecting a single column (returns a Series)
print("df1['Age']:\n", df1['Age'])
print("Type:", type(df1['Age']))
print()

# Selecting multiple columns (returns a DataFrame)
print("df1[['Name', 'GPA']]:\n", df1[['Name', 'GPA']])
print()

# .loc - label-based selection
print("df1.loc[0:2, ['Name', 'GPA']]:\n", df1.loc[0:2, ['Name', 'GPA']])
print()

# .iloc - integer-based selection
print("df1.iloc[1:4, 0:3]:\n", df1.iloc[1:4, 0:3])
print()

# Conditional selection
print("Rows where Age > 28:\n", df1[df1['Age'] > 28])
print()

# Multiple conditions
print("CS students with GPA > 3.7:\n", df1[(df1['Department'] == 'CS') & (df1['GPA'] > 3.7)])
```

### 2.4 Adding and Dropping Elements

```python
# Adding a new column
df1['Graduated'] = [False, True, True, False, True]
print("After adding 'Graduated' column:")
print(df1)
print()

# Adding a computed column
df1['Age_Group'] = df1['Age'].apply(lambda x: 'Young' if x < 30 else 'Senior')
print("After adding 'Age_Group' column:")
print(df1)
print()

# Adding a new row
new_row = pd.DataFrame({'Name': ['Frank'], 'Age': [29], 'Department': ['Physics'], 
                        'GPA': [3.4], 'Graduated': [False], 'Age_Group': ['Young']})
df1 = pd.concat([df1, new_row], ignore_index=True)
print("After adding new row:")
print(df1)
print()

# Dropping a column
df_dropped_col = df1.drop('Age_Group', axis=1)
print("After dropping 'Age_Group' column:")
print(df_dropped_col)
print()

# Dropping a row
df_dropped_row = df1.drop(5, axis=0)  # Drop row with index 5
print("After dropping row 5:")
print(df_dropped_row)
print()

# Dropping multiple columns/rows
df_clean = df1.drop(columns=['Age_Group', 'Graduated'])
df_clean = df_clean.drop(index=[4, 5])
print("After dropping multiple columns and rows:")
print(df_clean)
```

## Part 3: DataFrame Methods - Filtering, Sorting, Missing Values

### 3.1 Filtering Data

```python
# Create a sample dataset with some variety
np.random.seed(42)
data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace', 'Henry'],
    'Major': ['CS', 'Math', 'Physics', 'CS', 'Math', 'Physics', 'CS', 'Math'],
    'GPA': [3.8, 3.5, 3.9, 3.7, 3.6, 2.8, 3.9, 3.4],
    'Credits': [120, 90, 110, 130, 95, 60, 125, 85],
    'Research_Hours': [15, 10, 20, 12, 8, 5, 18, np.nan]
}

df = pd.DataFrame(data)
print("Sample dataset:")
print(df)
print()

# --- Filtering Methods ---

# 1. Boolean filtering
print("Students with GPA >= 3.7:")
high_gpa = df[df['GPA'] >= 3.7]
print(high_gpa)
print()

# 2. Using .query() method
print("CS students with Credits > 100 (using .query()):")
cs_advanced = df.query("Major == 'CS' and Credits > 100")
print(cs_advanced)
print()

# 3. Filtering with .isin()
print("Students in CS or Physics:")
stem_students = df[df['Major'].isin(['CS', 'Physics'])]
print(stem_students)
print()

# 4. Filtering with .between()
print("Students with GPA between 3.5 and 3.8:")
mid_gpa = df[df['GPA'].between(3.5, 3.8)]
print(mid_gpa)
print()

# 5. String methods for filtering
print("Students whose name starts with 'A' or 'B':")
ab_names = df[df['Student'].str.startswith(('A', 'B'))]
print(ab_names)
```

### 3.2 Sorting Data

```python
# Sort by a single column
print("Sorted by GPA (ascending):")
print(df.sort_values('GPA'))
print()

# Sort by a single column (descending)
print("Sorted by GPA (descending):")
print(df.sort_values('GPA', ascending=False))
print()

# Sort by multiple columns
print("Sorted by Major, then by GPA (descending):")
print(df.sort_values(['Major', 'GPA'], ascending=[True, False]))
print()

# Sort by index
df_shuffled = df.sample(frac=1, random_state=42)  # Shuffle rows
print("Shuffled DataFrame:")
print(df_shuffled)
print()
print("Sorted back by index:")
print(df_shuffled.sort_index())
print()

# Sort with NaN placement
print("Sorted by Research_Hours (NaN last):")
print(df.sort_values('Research_Hours', na_position='last'))
print()
print("Sorted by Research_Hours (NaN first):")
print(df.sort_values('Research_Hours', na_position='first'))
```

### 3.3 Handling Missing Values

```python
# Check for missing values
print("Missing values per column:")
print(df.isnull().sum())
print()

print("Total missing values:", df.isnull().sum().sum())
print()

# Check which rows have missing values
print("Rows with any missing values:")
print(df[df.isnull().any(axis=1)])
print()

# --- Strategies for handling missing values ---

# 1. Drop rows with missing values
df_dropna = df.dropna()
print("After dropna() (rows with NaN removed):")
print(df_dropna)
print()

# 2. Drop columns with missing values
df_drop_col = df.dropna(axis=1)
print("After dropna(axis=1) (columns with NaN removed):")
print(df_drop_col)
print()

# 3. Fill missing values with a constant
df_fill_zero = df.fillna(0)
print("After fillna(0):")
print(df_fill_zero)
print()

# 4. Fill with mean/median/mode
mean_hours = df['Research_Hours'].mean()
df_fill_mean = df.fillna({'Research_Hours': mean_hours})
print(f"After fillna with mean ({mean_hours:.2f}):")
print(df_fill_mean)
print()

# 5. Forward fill (use previous value)
df_ffill = df.fillna(method='ffill')
print("After forward fill (ffill):")
print(df_ffill)
print()

# 6. Backward fill (use next value)
df_bfill = df.fillna(method='bfill')
print("After backward fill (bfill):")
print(df_bfill)
print()

# 7. Interpolate (for numerical data)
df_interpolated = df.copy()
df_interpolated['Research_Hours'] = df_interpolated['Research_Hours'].interpolate()
print("After interpolation on Research_Hours:")
print(df_interpolated)
```

## Part 4: In-Class Example - Analyzing Graduate Research Data

**Scenario:** You are a graduate research assistant analyzing data from a multi-departmental study on graduate student performance and well-being.

**Dataset:** Simulated data for 20 graduate students across 4 departments.

**Tasks:**
1. Explore the dataset
2. Clean the data (handle missing values)
3. Filter and sort to find top performers
4. Compute department-level statistics
5. Identify students who need support

```python
# --- Setup: Create the in-class dataset ---
np.random.seed(2024)

n_students = 20

in_class_data = {
    'Student_ID': [f'S{i:03d}' for i in range(1, n_students + 1)],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace', 'Henry',
             'Ivy', 'Jack', 'Kate', 'Leo', 'Mia', 'Nathan', 'Olivia', 'Paul',
             'Quinn', 'Rachel', 'Sam', 'Tina'],
    'Department': np.random.choice(['CS', 'Math', 'Physics', 'Biology'], n_students),
    'Year': np.random.choice([1, 2, 3, 4, 5], n_students),
    'GPA': np.round(np.random.normal(3.5, 0.4, n_students), 2),
    'Research_Hours': np.random.choice([np.nan, 5, 10, 15, 20, 25, 30], n_students),
    'Publications': np.random.choice([0, 0, 0, 1, 1, 2, 3], n_students),
    'Stress_Level': np.random.choice([1, 2, 3, 4, 5], n_students),  # 1=low, 5=high
    'Advisor_Satisfaction': np.round(np.random.uniform(2.0, 5.0, n_students), 1)
}

research_df = pd.DataFrame(in_class_data)

print("=== GRADUATE RESEARCH DATASET ===")
print(research_df)
print()
print(f"Shape: {research_df.shape}")
print(f"Columns: {list(research_df.columns)}")
```

### Task 1: Initial Exploration

```python
# Quick overview
# print("=== DATA OVERVIEW ===")
print(research_df.info())
print()

print("=== DESCRIPTIVE STATISTICS ===")
print(research_df.describe())
print()
'''
print("=== MISSING VALUES ===")
print(research_df.isnull().sum())
print()

# print("=== DEPARTMENT DISTRIBUTION ===")
# print(research_df['Department'].value_counts())
# print()
'''
print("=== YEAR DISTRIBUTION ===")
print(research_df['Year'].value_counts().sort_index())
```

### Task 2: Data Cleaning

```python
# Make a copy for cleaning
df_clean = research_df.copy()

# Fill missing Research_Hours with department median
dept_median = df_clean.groupby('Department')['Research_Hours'].transform('median')
df_clean['Research_Hours'] = df_clean['Research_Hours'].fillna(dept_median)

# Verify no missing values remain
print("Missing values after cleaning:")
print(df_clean.isnull().sum())
print()

print("Cleaned dataset:")
print(df_clean.head(10))
```

### Task 3: Find Top Performers

```python
# Top 5 students by GPA
print("=== TOP 5 STUDENTS BY GPA ===")
top_gpa = df_clean.nlargest(5, 'GPA')[['Name', 'Department', 'GPA', 'Publications']]
print(top_gpa)
print()

# Top 5 students by Publications (then GPA as tiebreaker)
print("=== TOP 5 STUDENTS BY PUBLICATIONS ===")
top_pubs = df_clean.sort_values(['Publications', 'GPA'], ascending=[False, False]).head(5)
print(top_pubs[['Name', 'Department', 'Publications', 'GPA']])
print()

# Students with high research hours AND high GPA
print("=== HIGH RESEARCH + HIGH GPA STUDENTS ===")
high_perf = df_clean[(df_clean['Research_Hours'] >= 20) & (df_clean['GPA'] >= 3.5)]
print(high_perf[['Name', 'Department', 'GPA', 'Research_Hours', 'Publications']])
```

### Task 4: Department-Level Statistics

```python
# Group by department and compute statistics
print("=== DEPARTMENT STATISTICS ===")
dept_stats = df_clean.groupby('Department').agg({
    'GPA': ['mean', 'min', 'max', 'std'],
    'Research_Hours': ['mean', 'sum'],
    'Publications': ['mean', 'sum'],
    'Stress_Level': 'mean',
    'Advisor_Satisfaction': 'mean'
}).round(2)

print(dept_stats)
print()

# Simpler aggregation
print("=== AVERAGE GPA BY DEPARTMENT ===")
avg_gpa = df_clean.groupby('Department')['GPA'].mean().sort_values(ascending=False)
print(avg_gpa)
print()

# Count students by department and year
print("=== STUDENT COUNT BY DEPARTMENT AND YEAR ===")
pivot = df_clean.pivot_table(index='Department', columns='Year', values='Student_ID', aggfunc='count', fill_value=0)
print(pivot)
```

### Task 5: Identify Students Needing Support

```python
# Students who might need support:
# - Low GPA (< 3.0)
# - High stress (>= 4)
# - Low advisor satisfaction (< 3.0)
# - Low research hours (< 10) in year 3+

print("=== STUDENTS NEEDING SUPPORT ===")

# Criteria 1: Low GPA
low_gpa = df_clean[df_clean['GPA'] < 3.0]
print("1. Low GPA (< 3.0):")
print(low_gpa[['Name', 'Department', 'GPA', 'Year']])
print()

# Criteria 2: High stress
high_stress = df_clean[df_clean['Stress_Level'] >= 4]
print("2. High Stress (>= 4):")
print(high_stress[['Name', 'Department', 'Stress_Level', 'Advisor_Satisfaction']])
print()

# Criteria 3: Low advisor satisfaction
low_sat = df_clean[df_clean['Advisor_Satisfaction'] < 3.0]
print("3. Low Advisor Satisfaction (< 3.0):")
print(low_sat[['Name', 'Department', 'Advisor_Satisfaction', 'GPA']])
print()

# Criteria 4: Low research hours in year 3+
low_research = df_clean[(df_clean['Year'] >= 3) & (df_clean['Research_Hours'] < 10)]
print("4. Low Research Hours (< 10) in Year 3+:")
print(low_research[['Name', 'Department', 'Year', 'Research_Hours']])
print()

# Combined: Any at-risk student
at_risk = df_clean[
    (df_clean['GPA'] < 3.0) |
    (df_clean['Stress_Level'] >= 4) |
    (df_clean['Advisor_Satisfaction'] < 3.0) |
    ((df_clean['Year'] >= 3) & (df_clean['Research_Hours'] < 10))
]
print("=== ALL AT-RISK STUDENTS ===")
print(at_risk[['Name', 'Department', 'GPA', 'Stress_Level', 'Advisor_Satisfaction', 'Year', 'Research_Hours']])
print(f"\nTotal at-risk students: {len(at_risk)} / {len(df_clean)}")
```

### Bonus: Creating a Summary Report

```python
# Create a comprehensive summary report
print("=" * 60)
print("GRADUATE STUDENT RESEARCH SUMMARY REPORT")
print("=" * 60)
print()

print(f"Total Students: {len(df_clean)}")
print(f"Departments: {', '.join(df_clean['Department'].unique())}")
print(f"Average GPA: {df_clean['GPA'].mean():.2f}")
print(f"Average Research Hours: {df_clean['Research_Hours'].mean():.1f}")
print(f"Total Publications: {df_clean['Publications'].sum()}")
print(f"Average Stress Level: {df_clean['Stress_Level'].mean():.2f} / 5.0")
print(f"At-Risk Students: {len(at_risk)} ({len(at_risk)/len(df_clean)*100:.1f}%)")
print()

print("Department Rankings (by Average GPA):")
ranking = df_clean.groupby('Department')['GPA'].mean().sort_values(ascending=False)
for i, (dept, gpa) in enumerate(ranking.items(), 1):
    print(f"  {i}. {dept}: {gpa:.2f}")

print()
print("=" * 60)
```

## Summary: Key Pandas Concepts Covered

| Concept | Methods | Description |
|---------|---------|-------------|
| **Series Creation** | `pd.Series()`, from dict/list | 1D labeled array |
| **DataFrame Creation** | `pd.DataFrame()`, from dict/list/NumPy | 2D labeled data |
| **Index** | `.index`, `.set_index()`, `.reset_index()` | Row labels |
| **Slicing** | `.loc[]`, `.iloc[]` | Label vs position-based |
| **Adding Elements** | Direct assignment, `.concat()` | New rows/columns |
| **Dropping Elements** | `.drop()` | Remove rows/columns |
| **Filtering** | Boolean indexing, `.query()`, `.isin()`, `.between()` | Select subsets |
| **Sorting** | `.sort_values()`, `.sort_index()` | Order data |
| **Missing Values** | `.isnull()`, `.dropna()`, `.fillna()`, `.interpolate()` | Handle NaNs |
| **Grouping** | `.groupby()`, `.agg()`, `.pivot_table()` | Aggregate data |
